```python
import os
import pandas as pd
from tqdm import tqdm
```

```python
main_dir = '/Users/stevie/repos/lingo_kit_data/vocabulary/dataframes'
ignore = ['AUX']
count = 0
pos_folders = os.listdir(main_dir)
for pos_folder in tqdm(pos_folders):
    if pos_folder in ignore:
        print("Ignoring", pos_folder)
        continue
    pos_dir = os.path.join(main_dir, pos_folder)
    for lemma_file in os.listdir(pos_dir):
        lemma_path = os.path.join(pos_dir, lemma_file)
        df = pd.read_csv(lemma_path, sep='\t')
        count += len(df)
print("Total words across all parts of speech:", count)
```
