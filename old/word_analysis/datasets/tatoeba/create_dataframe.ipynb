```python
from tqdm import tqdm

import pandas as pd
```

```python
path = '/Users/stevie/repos/lingo_kit_combined/lingo_kit_data/word_analysis/datasets/tatoeba/Sentence pairs in Italian-English - 2025-10-16.tsv'
new_path = '/Users/stevie/repos/lingo_kit_combined/lingo_kit_data/word_analysis/datasets/tatoeba/dataframe.tsv'

# read the tsv file line by line, if there are too many tabs, ignore the line
# write to a csv file
# count error lines
error_lines = 0
with open(new_path, 'w') as f:
    with open(path, 'r') as infile:
        all_lines = infile.readlines()
        for line_i, line in tqdm(enumerate(all_lines), total=len(all_lines)):
            if line_i == 597259 or line_i == 0:
                continue
            if line.count('\t') != 3:
                error_lines += 1
                continue
            f.write(line.replace('"', ''))
print(f'Error lines: {error_lines}')
```

```python
df = pd.read_csv(new_path, sep='\t', names=['id_it', 'text_it', 'id_en', 'text_en'])
len(df), df.columns
```

```python
# hash italian text to create a unique id
import uuid
def hash_text(text):
    return str(uuid.uuid5(uuid.NAMESPACE_DNS, text))
df['hash'] = df['text_it'].apply(hash_text)
df.columns
```

```python
df.drop(columns=['id_it', 'id_en'], inplace=True)
```

```python
print(df['text_it'].duplicated().sum())
print(df['hash'].duplicated().sum())
```

```python
print(len(df))
df.drop_duplicates(subset=['hash'], inplace=True)
print(len(df))
```

```python
# shuffle the dataframe
df.sample(frac=1)
```

```python
df.to_csv(new_path, sep='\t', index=False)
```
