```python
import pandas as pd
import os
```

```python
def get_all_tsvs_rec(path):
    paths = []
    if os.path.isdir(path):
        for entry in os.listdir(path):
            full_path = os.path.join(path, entry)
            paths.extend(get_all_tsvs_rec(full_path))
    elif path.endswith('.tsv'):
        paths.append(path)
    return paths
# adj_paths = get_all_tsvs_rec('dataframes/ADJ')
# noun_paths = get_all_tsvs_rec('dataframes/NOUN')
# paths = adj_paths + noun_paths
# print(paths)
paths = get_all_tsvs_rec('dataframes')
```

```python
len(paths)
```

```python
cols = [
    'text', 'lemma', 'pos', 'translation_en', 'token_count', 'token_pct', 'group_count',
    'group_pct', 'xpos', 'deprel', 'feats', 'token_hash', 'group_hash',
    'group', 'sentence_1_it', 'sentence_1_en', 'sentence_2_it',
    'sentence_2_en', 'sentence_3_it', 'sentence_3_en'
]
for path in paths:
    df = pd.read_csv(path, sep='\t')
    assert(set(cols) == set(df.columns))
    df = df[cols]
    for i, row in df.iterrows():
        for col in cols:
            if type(row[col]) == str:
                df.loc[i, col] = row[col].strip()
    df.to_csv(path, sep='\t', index=False)
```
