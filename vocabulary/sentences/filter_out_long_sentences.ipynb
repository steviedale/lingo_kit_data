```python
import pandas as pd
```

```python
df = pd.read_csv('sentences/sentence_dataframe.tsv', sep='\t')
len(df), df.columns
```

```python
df['text_it_length'] = df['text_it'].str.len()
df['text_en_length'] = df['text_en'].str.len()
```

```python
import numpy as np
import matplotlib.pyplot as plt
bins = np.linspace(0, 125, 125)
df['text_it_length'].hist(bins=bins, alpha=0.5, label='Italian', color='orange')
df['text_en_length'].hist(bins=bins, alpha=0.5, label='English', color='blue')
plt.legend()
```

```python
df['text_length_diff'] = df['text_it_length'] - df['text_en_length']
bins = np.linspace(-50, 50, 101)
df['text_length_diff'].hist(bins=bins)
# plot a vertical line at x=0, dashed
plt.axvline(x=0, color='red', linestyle='dashed')
```

```python
N = 50
long_df = df[
    (df['text_it_length'] > N) &
    (df['text_it_length'] < N+5)
]
for i in range(5):
    print(long_df.iloc[i]['text_it'])
    print(long_df.iloc[i]['text_en'])
    print('---')
```

```python
N = 50
short_df = df[
    (df['text_it_length'] <= N) &
    (df['text_en_length'] <= N)
]
short_df['text_it_length'].hist(alpha=0.5, label='Italian', color='orange')
short_df['text_en_length'].hist(alpha=0.5, label='English', color='blue')
```

```python
short_df.columns
```

```python
len(short_df)
```

```python
short_df.drop(columns=['text_length_diff'], inplace=True)
short_df.columns
```

```python
short_df.to_csv('sentences/short_lte50_sentence_dataframe.tsv', sep='\t', index=False)
```

```python
import os
# lets break this file into chunks
chunk_size = 100000
if len(short_df) % chunk_size == 0:
    n_chunks = len(short_df) // chunk_size
else: 
    n_chunks = len(short_df) // chunk_size + 1
for chunk_i in range(n_chunks):
    start_i = chunk_i * chunk_size
    end_i = min((chunk_i + 1) * chunk_size, len(short_df))
    chunk_df = short_df.iloc[start_i:end_i]
    save_path = f'sentences/short_lte{N}_sentences/chunk_{chunk_i+1:02d}.tsv'
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    chunk_df.to_csv(save_path, sep='\t', index=False)
```
