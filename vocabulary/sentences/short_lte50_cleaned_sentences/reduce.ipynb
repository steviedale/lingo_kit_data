```python
# let's break up these .tsv files into chunks of 10,000 (instead of 100,000)
```

```python
import os
import pandas as pd
```

```python
big_dir = 'big_chunks'
os.listdir(big_dir)
```

```python
small_dir = 'smaller_chunks'
os.makedirs(small_dir, exist_ok=True)
small_chunk_counter = 0
for filename in sorted(os.listdir(big_dir)):
    if not filename.endswith('.tsv'):
        continue

    filepath = os.path.join(big_dir, filename)
    print(f'Processing {filepath}...')

    df = pd.read_csv(filepath, sep='\t')

    chunk_size = 10000
    num_chunks = (len(df) + chunk_size - 1) // chunk_size

    for i in range(num_chunks):
        chunk_df = df[i*chunk_size:(i+1)*chunk_size]
        chunk_filename = f'chunk_{small_chunk_counter:04d}.tsv'
        chunk_filepath = os.path.join(small_dir, chunk_filename)
        chunk_df.to_csv(chunk_filepath, sep='\t', index=False)
        small_chunk_counter += 1
        print(f'  Wrote {chunk_filepath} with {len(chunk_df)} rows.')
```
