```python
import os
import pandas as pd
```

```python
from tqdm import tqdm
sentence_cols = [
    'sentence_1_it',
    'sentence_1_en',
    'sentence_2_it',
    'sentence_2_en',
    'sentence_3_it',
    'sentence_3_en',
]
df_dir = '/Users/stevie/repos/lingo_kit_super/lingo_kit_data/vocabulary/dataframes'
for pos in os.listdir(df_dir):
    print(pos)
    pos_dir = os.path.join(df_dir, pos)
    for file in tqdm(os.listdir(pos_dir)):
        path = os.path.join(pos_dir, file)
        df = pd.read_csv(path, sep='\t')
        df.drop(columns=sentence_cols, inplace=True)
        df.to_csv(path, sep='\t', index=False)
```
