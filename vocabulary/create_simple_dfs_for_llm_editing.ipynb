```python
import os
import pandas as pd
```

```python
def get_all_tsv_files(directory):
    """Recursively get all .tsv files in the given directory."""
    tsv_files = []
    for root, _, files in os.walk(directory):
        for file in files:
            if file.endswith('.tsv'):
                tsv_files.append(os.path.join(root, file))
    return tsv_files
old_dir = '/Users/stevie/repos/lingo_kit_super/lingo_kit_data/vocabulary/dataframes/'
new_dir = '/Users/stevie/repos/lingo_kit_super/lingo_kit_data/vocabulary/dataframes_cleaned/'
files = get_all_tsv_files(old_dir)
files
```

```python
keep_columns = ['text', 'translation_en', 'parts_lemma', 'parts_pos', 'feats']
for file in files:
    assert(old_dir in file)
    new_path = file.replace(old_dir, new_dir)
    print(new_path)
    assert(new_dir in new_path)
    assert(old_dir not in new_path)
    df = pd.read_csv(file, sep='\t')
    # print(df.columns)
    # break
    if not os.path.exists(os.path.dirname(new_path)):
        os.makedirs(os.path.dirname(new_path))
    df[keep_columns].to_csv(new_path, sep='\t', index=False)
```
