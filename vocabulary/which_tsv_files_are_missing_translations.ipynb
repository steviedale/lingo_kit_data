```python
def get_tsv_files_rec(dir):
    tsv_files = []
    for root, dirs, files in os.walk(dir):
        for file in files:
            if file.endswith('.tsv'):
                tsv_files.append(os.path.join(root, file))
    return tsv_files
```

```python
missing_lemmas = []
missing_pos = set()
missing_files = []
tsv_files = get_tsv_files_rec(dataframe_dir)
for file_path in tsv_files:
    if file_path.endswith('.tsv'):
        df = pd.read_csv(file_path, sep='\t')
        missing_translation_count = df['translation_en'].isnull().sum()
        missing_pron_count = df['pronunciation'].isnull().sum()
        pos = os.path.basename(os.path.dirname(file_path))
        lemma = os.path.splitext(os.path.basename(file_path))[0]
        if missing_translation_count > 0 or missing_pron_count > 0:
            missing_lemmas.append((lemma, pos, int(missing_translation_count), int(missing_pron_count)))
            missing_pos.add(pos)
            missing_files.append(file_path)
```

```python
start_letters = set([x[0][0] for x in missing_lemmas])
start_letters
```
