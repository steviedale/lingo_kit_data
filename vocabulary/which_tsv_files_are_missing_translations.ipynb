```python
import os
import pandas as pd
```

```python
dataframe_dir = 'dataframes'
```

```python
missing_pos = set()
for pos in os.listdir(dataframe_dir):
    pos_path = os.path.join(dataframe_dir, pos)
    for file in os.listdir(pos_path):
        if file.endswith('.tsv'):
            file_path = os.path.join(pos_path, file)
            df = pd.read_csv(file_path, sep='\t')
            missing_translation_count = df['translation_en'].isnull().sum()
            missing_pron_count = df['pronunciation'].isnull().sum()
            if missing_translation_count > 0 or missing_pron_count > 0:
                missing_pos.add(pos)
                print(f"File: {file_path}")
                if missing_translation_count > 0:
                    print(f"  Missing translations: {missing_translation_count}")
                if missing_pron_count > 0:
                    print(f"  Missing pronunciations: {missing_pron_count}")
if missing_pos:
    print("\nParts of speech with missing data:")
    for pos in missing_pos:
        print(f"- {pos}")
```
