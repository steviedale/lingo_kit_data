```python
import os
import pandas as pd
```

```python
# dataframe_dir = 'bigrams'
dataframe_dir = 'trigrams'
```

```python
missing_translation_counts = {}
for chunk_file in sorted(os.listdir(dataframe_dir)):
    chunk_path = os.path.join(dataframe_dir, chunk_file)
    if chunk_path.endswith('.tsv'):
        df = pd.read_csv(chunk_path, sep='\t')
        missing_translation_count = df['translation_en'].isnull().sum()
        if missing_translation_count > 0:
            missing_translation_counts[chunk_file] = missing_translation_count
            print(f"File: {chunk_file}, Missing translations: {missing_translation_count}")
```
