```python
import os
import pandas as pd
```

```python
# save_dir = 'bigrams'
# df = pd.read_csv('gram2.tsv', sep='\t')

save_dir = 'trigrams'
df = pd.read_csv('gram3.tsv', sep='\t')

len(df), df.columns
```

```python
df.sort_values('count', inplace=True, ascending=False)
df.head()
```

```python
df['translation_en'].notnull().sum()
```

```python
os.makedirs(save_dir, exist_ok=True)
```

```python
chunk_size = 100
n_chunks = len(df) // chunk_size
remainder = len(df) % chunk_size
if remainder > 0:
    n_chunks += 1
for chunk_i in range(n_chunks):
    start_i = chunk_i * chunk_size
    end_i = min((chunk_i + 1) * chunk_size, len(df))
    chunk_df = df[start_i:end_i]
    chunk_path = os.path.join(save_dir, f'chunk_{chunk_i:03d}.tsv')
    nan_count = chunk_df['translation_en'].isna().sum()
    if nan_count == 0:
        print('completed chunk', chunk_i)
    elif nan_count > 0 and nan_count < end_i - start_i:
        print('partial chunk', chunk_i, 'with', nan_count, 'missing translations')
    chunk_df.to_csv(chunk_path, sep='\t', index=False)
    # print(f'Saved chunk {chunk_i} to {chunk_path}')
```
