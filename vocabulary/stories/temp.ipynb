```python
import os
import pandas as pd
```

```python
data = {'italian': [], 'english': []}
with open('sara_perche_ti_amo.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()
    # first line is the italian sentence
    # second line is the english translation
    # third line is empty
    for i, line in enumerate(lines):
        if i % 3 == 0:
            data['italian'].append(line.strip())
        elif i % 3 == 1:
            data['english'].append(line.strip())
df = pd.DataFrame(data)
```
